## `nb-cli` notes

### Setting up

Install `jupyter-collaboration` after installing `jupyterlab` in the environment to use the `connect` mode

**Note**: Use the [forked version](https://github.com/wingedrasengan927/nb-cli) of `nb-cli`.

**Q**: How to connect `nb-cli` to a live jupyter lab server?

**A**: Ensure `jupyter-collaboration` is installed and run the following command after replacing the `server` and `token` parameters.
```bash
nb connect --server http://localhost:8889 --token e1dcef70fa68335223680e2d5487597a3fcfa70c66435423
```

---

It elegantly reads output cells as well! To read a notebook, run `nb read <notebook-name>`
